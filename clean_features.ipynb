```python
import json
import pandas as pd
```

```python
pre_selected_features_filename = "fe_eclipse_tsfresh_raw_CHI_2000.json"

with open(pre_selected_features_filename, "r") as fp:
    selected_features_json = json.load(fp)
    fe_selected_features = selected_features_json['tsfresh_column_names']
```

```python
list(fe_selected_features.items())[0][1]['matrix_profile']
```

```python
fe_selected_features.keys()
```

```python
clean_features = dict()

# remove matrix_profile
for k,v in fe_selected_features.items():
    
    assert isinstance(v, dict), f"Expected dict, got {type(v)}"
    # matrix_profile in v, then remove matrix_profile and add to clean_features
    if 'matrix_profile' in v:
        v.pop('matrix_profile')
        clean_features[k] = v
```

```python
clean_features
```

```python
with open("fe_eclipse_tsfresh_raw_CHI_2000_clean.json", "w") as fp:
    json.dump(clean_features, fp)  # write the cleaned features to a new
```
